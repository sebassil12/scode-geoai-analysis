```python
import matplotlib.pyplot as plt
import seaborn as sns

# Datos de importancia de características
features = ["elevation", "NDVI_2022", "NDVI_mean_3yr", "NDVI_2023", "Red_homogeneity"]
importance = [0.169114, 0.073313, 0.069337, 0.062558, 0.050714]

# Estilo general
sns.set(style="whitegrid", font_scale=1.2)

# Crear figura
plt.figure(figsize=(10, 6))

# Paleta estética
colors = sns.color_palette("viridis", len(features))

# Gráfico de barras
bars = plt.bar(features, importance, color=colors)

# Etiquetas encima de cada barra
for bar in bars:
    height = bar.get_height()
    plt.annotate(f"{height:.3f}", 
                 xy=(bar.get_x() + bar.get_width() / 2, height),
                 xytext=(0, 8),
                 textcoords="offset points",
                 ha='center', va='bottom', fontsize=11, color="black")

# Títulos
plt.title("Top 5 Características por Importancia del Modelo", fontsize=16, fontweight='bold')
plt.ylabel("Importancia", fontsize=14)
plt.xticks(rotation=30, ha='right')

# Ajuste final
plt.tight_layout()
plt.show()
```
