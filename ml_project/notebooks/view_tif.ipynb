```python
import rasterio
import numpy as np

# Ruta al archivo GeoTIFF
geotiff_path = 'deforestation_map_Q3_2025.tif'

# El área de cada píxel es de 30m x 30m, según la escala de exportación de GEE.
pixel_area_m2 = 30 * 30

with rasterio.open(geotiff_path) as src:
    # Leer la primera banda de la imagen
    band1 = src.read(1)

# Filtrar solo los píxeles que pertenecen a nuestras clases (0, 1, o 2)
# Esto evita contar valores nulos o de fondo si los hubiera.
valid_pixels = band1[band1 >= 0]

# Obtener las clases únicas y el conteo de píxeles para cada una
unique_classes, pixel_counts = np.unique(valid_pixels, return_counts=True)

# Crear un diccionario para mapear la clase a su nombre
class_names = {
    0: "Otras Tierras",
    1: "Bosque",
    2: "Deforestado"
}

print("--- Resultados del Análisis de Área (Corregido) ---")
print(f"Área por píxel: {pixel_area_m2} m² (escala de 30x30m)")
print("-" * 45)
print("Conteo de píxeles por clase:", dict(zip(unique_classes, pixel_counts)))
print("-" * 45)

total_area_ha = 0

# Iterar sobre los resultados para calcular y mostrar el área
for cls, count in zip(unique_classes, pixel_counts):
    # Convertir el conteo de píxeles a hectáreas
    area_ha = (count * pixel_area_m2) / 10000
    total_area_ha += area_ha
    
    class_name = class_names.get(cls, f"Clase Desconocida ({cls})")
    print(f"Clase '{class_name}': {area_ha:,.2f} hectáreas")

print("-" * 45)
print(f"Área Total Analizada: {total_area_ha:,.2f} hectáreas")
print("-----------------------------------------------")
```
